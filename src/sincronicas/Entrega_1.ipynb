### 2. Usar la primera clase para obtener el DataFrame desde Hugging Face.

Configurar el dataset y la salida

### 3. Aplicar una transformación mínima al DataFrame cargado.

```python
# 1. Eliminar la columna redundante 'idx' porque es la misma que el índice del DataFrame
df = df.drop(columns=["idx"])
# Verificación de la lista de columnas
print(f"Verificando que la columna 'idx' ha sido eliminada:\n{df.columns}")
```

```python
# Calculando el Índice de Masa Corporal (IMC)
df.loc[:, "bmi"] = df["weight"] / ((df["height"] / 100) ** 2)
print("Se ha calculado el Índice de Masa Corporal (IMC) y se ha agregado como una nueva columna 'bmi'.")
df.head(2)
```

```python
# Convirtiendo a variable ordinal de la variable objetivo ('risk_level')
risk_mapping = {"low": 0, "medium": 1, "high": 2}
df.loc[:, "risk_level_encoded"] = df["risk_level"].map(risk_mapping)

print("Se ha convertido la variable objetivo 'risk_level' a una variable ordinal 'risk_level_encoded'.")
df.head(2)
```

###  Resultado y explicación de las transformaciones



1. **Eliminación de `idx`:** Se removió la columna identificadora por ser redundante con el índice interno de Pandas ($i \in \{0, 1, \dots, n-1\}$).

2. **Creación de `bmi` ($\text{IMC}$):** Se combinaron el peso ($\text{weight}$) y la estatura ($\text{height}$) en una nueva variable  usando la fórmula del Índice de Masa Corporal:

   $$\text{BMI} = \frac{\text{weight}_{\text{kg}}}{\left(\frac{\text{height}_{\text{cm}}}{100}\right)^2}$$
